## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
print("""
Methodology:

The analysis uses content-performance features to identify groups of pages with similar observed patterns.

Features used:
- impressions_90d
- clicks_90d
- sessions_90d
- avg_position
- ctr
- engagement_rate
- scroll_rate
- word_count
- content_age_days
- days_since_last_update

Rows with missing values in the modeling features are excluded.

The numeric features are standardized using StandardScaler so that variables with different scales do not dominate the clustering.

K-Means clustering is then applied with k=2, matching the number of clusters used in the earlier analysis.

Cluster profiles are examined using the original feature values. The lower-traffic and lower-engagement cluster is mapped to "review_and_refresh", while the stronger observed pattern is mapped to "monitor".

The action mapping is a human-defined interpretation of the observed clusters; the cluster labels themselves do not have an intrinsic business meaning.

The final queue is ranked using action priority and impressions_90d.

No future outcome is used to define the action labels, and the recommendations are intended for human review rather than automated content changes.
""")
```

```python
print("""
Limitations:

The clustering identifies groups with similar observed performance patterns; it does not establish that a page is objectively good or bad.

The analysis is descriptive and does not prove that refreshing a page will improve future performance.

Cluster assignments depend on the selected features, preprocessing choices, and the number of clusters.

The action mapping is a human interpretation of the observed cluster profiles and should not be treated as an automatic decision rule.

The results can change when the dataset, time window, content mix, or feature definitions change.

The recommendations are therefore intended to prioritize human review rather than automatically change or delete content.
""")
```
